```python
# Import necessary libraries
import pandas as pd
import os
```

```python
# Define file paths
real_file_path = '../utils/feature-extractor/results/real_features_20250624_211136.csv'  # Change this path as needed
fake_file_path = '../utils/feature-extractor/results/fake_features_20250624_223258.csv'  # Change this path as needed

# Load CSV files
real_df = pd.read_csv(real_file_path)
fake_df = pd.read_csv(fake_file_path)

real_df.info()
fake_df.info()
```

```python
# Drop half the rows from each
real_df = real_df.sample(frac=0.5, random_state=420).reset_index(drop=True)
fake_df = fake_df.sample(frac=0.5, random_state=420).reset_index(drop=True)

real_df.info()
fake_df.info()
```

```python
# Add 'class' column
real_df['class'] = 0
fake_df['class'] = 1

# Combine the dataframes
combined_df = pd.concat([real_df, fake_df], ignore_index=True)

# Save to disk
output_path = 'combined_dataset.csv'
combined_df.to_csv(output_path, index=False)

print(f"Combined dataset saved to: {os.path.abspath(output_path)}")
```
